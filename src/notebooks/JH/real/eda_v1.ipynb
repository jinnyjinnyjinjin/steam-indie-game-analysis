### [Step 1] 분석 데이터 로드 및 기본 구조 파악
분석에 필요한 주요 데이터셋(`graded`, `review_summary`, `reviews`)을 불러오고, 각 파일의 컬럼명, 데이터 타입, 결측치 여부 및 샘플 데이터를 확인하여 데이터 명세를 점검합니다.

```python
import pandas as pd
from IPython.display import display

# 스크린샷 기반 정확한 상대 경로 세팅
base_path = '../../../../data/preprocessed/'

# 스크린샷에 있는 실제 파일명 목록 (_2 제거)
files = [
    "steam_indie_games_graded.csv",
    "steam_indie_games.csv",
    "steam_indie_review_histogram.csv",
    "steam_indie_review_summary.csv",
    "steam_indie_reviews.csv"
]

for file_name in files:
    file_path = f"{base_path}{file_name}"
    try:
        # 데이터 불러오기
        df = pd.read_csv(file_path)
        
        # 명세서(Specification) 데이터프레임 생성
        spec_df = pd.DataFrame({
            'Column Name': df.columns,
            'Data Type': df.dtypes.astype(str).values,
            'Non-Null Count': df.notnull().sum().values,
            'Sample Value': df.iloc[0].astype(str).values if not df.empty else None
        })
        
        print(f"========== {file_name} ==========")
        display(spec_df) # df 형태로 깔끔하게 출력
        print("\n")
        
    except FileNotFoundError:
        print(f"Error: '{file_path}' 파일을 찾을 수 없습니다. 경로를 다시 확인해주세요.")
    except Exception as e:
        print(f"Error reading '{file_path}': {e}")
```

```python
import pandas as pd
import numpy as np

# 분석할 파일 목록
files = [
    "steam_indie_games.csv",
    "steam_indie_review_histogram.csv",
    "steam_indie_review_summary.csv",
    "steam_indie_reviews.csv",
    "steam_indie_tags.csv",
    "steamspy_indie_games.csv"
]

def check_data_health(file_path):
    print(f"\n{'='*70}")
    print(f"🔍 데이터 건강검진 리포트: {file_path}")
    print(f"{'='*70}")
    
    try:
        # 데이터 로드
        df = pd.read_csv(file_path)
        
        # 1. 기본 구조 확인
        print(f"[1] 규모: {df.shape[0]:,} 행 x {df.shape[1]} 열")
        
        # 2. 결측치(Missing Values) 확인
        null_info = df.isnull().sum()
        if null_info.sum() > 0:
            print("\n[2] 결측치 발견 (주의 필요):")
            print(null_info[null_info > 0])
        else:
            print("\n[2] 결측치: 없음 (깨끗함)")
            
        # 3. 중복 데이터 확인
        dup_count = df.duplicated().sum()
        print(f"\n[3] 중복 데이터: {dup_count}건")
        
        # 4. 수치형 데이터 이상치(Outlier) 후보 확인
        # min이 0이거나 음수인 경우, max가 비정상적으로 큰 경우를 체크합니다.
        num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
        if num_cols:
            print("\n[4] 수치 데이터 분포 (이상치 판단 근거):")
            stats = df[num_cols].describe().loc[['min', '50%', 'max']]
            print(stats.to_string())
            
        # 5. 핵심 키(Key) 고유성 확인
        if 'appid' in df.columns:
            print(f"\n[5] 고유 AppID 수: {df['appid'].nunique():,}개")
            
    except Exception as e:
        print(f"❌ 파일 읽기 중 에러 발생: {e}")

# 실행
for f in files:
    check_data_health(f)
```

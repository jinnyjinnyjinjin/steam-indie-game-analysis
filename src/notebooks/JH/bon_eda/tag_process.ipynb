```python
import pandas as pd
import os

# 1. 경로 설정 (이미지 구조 기준)
# 현재 위치: src/notebooks/JH/bon_eda/tag_process.ipynb
# 원본 데이터 위치: data/raw/
# 저장 데이터 위치: data/processed/

# 상위로 4번 올라가서 data 폴더로 진입
raw_data_path = "../../../../data/raw"
processed_path = "../../../../data/processed"

# 저장할 폴더가 없으면 생성
if not os.path.exists(processed_path):
    os.makedirs(processed_path)
    print(f"폴더 생성 완료: {processed_path}")

# 2. 데이터 로드 (data/raw/ 에서 읽기)
try:
    df_tags = pd.read_csv(os.path.join(raw_data_path, 'steam_indie_tags.csv'))
    df_games = pd.read_csv(os.path.join(raw_data_path, 'steam_indie_games.csv'))
except FileNotFoundError as e:
    print(f"파일을 찾을 수 없습니다. 경로를 확인하세요.\n{e}")

# 3. 컬럼 제거 실행
drop_cols = ['owners', 'price', 'positive', 'negative', 'name', 'updated_at']
df_tags_cleaned = df_tags.drop(columns=[col for col in drop_cols if col in df_tags.columns])

# 4. 결과 확인
print(f"남은 컬럼: {df_tags_cleaned.columns.tolist()}")

# 5. 저장 (data/processed/ 에 저장)
save_file_path = os.path.join(processed_path, 'steam_indie_tags_clean.csv')
df_tags_cleaned.to_csv(save_file_path, index=False)
```
